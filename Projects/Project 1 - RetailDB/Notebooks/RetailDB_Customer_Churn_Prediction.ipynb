Mapping RDDs as key-rows

```python
categories_rdd_mapped = categories_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))  # Mapping by category_id
customers_rdd_mapped = customers_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by customer_id
departments_rdd_mapped = departments_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))  # Mapping by department_id
orders_rdd_mapped = orders_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by order_id
order_items_rdd_mapped = order_items_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[1]), x))  # Mapping by order_item_id (use x[1] for order_id)
products_rdd_mapped = products_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by product_id
```

Mapping columns onto split RDD rows and converting them to dataframes

```python
from pyspark.sql import Row

# Split RDD and map it to Row objects
customers_rdd_mapped = customers_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    cust_id=int(x[0]),
    fname=x[1],
    lname=x[2],
    unknown_1=x[3],
    unknown_2=x[4],
    address=x[5],
    city=x[6],
    state=x[7],
    value=float(x[8])
))

# Convert to DataFrame
customers_df = customers_rdd_mapped.toDF()
customers_df.show()
```

## **Performing Customer Churn Prediction using Logistic Regression & Decision Tree**

Cleaning Customers_df

Setting up categories, departments and products as required

```python
customer_order_df = customer_order_df.drop("o_id", "c_id")
customer_order_df.show()
```

```python
# Select relevant features for customer churn prediction
customer_churn_df = customer_order_df.select(
    "cust_id",
    "p_id",
    "order_id",
    "cust_value",
    "order_value",
    "item_value",
    "quantity",
    "unit_price",
    "year",
    "month",
    "state",
    "city",
    "status"
)
customer_churn_df.show()
```

```python
# Checking all available years to set churn threshold
customer_churn_df.select("year").distinct().show()
```

Since our dataset spans over 2 years, we shall assume 3 months as the churn threshold.

```python
# Finding the date of the latest order to begin threshod checking from.

latest_order_date = orders_df.select(F.max(F.to_date("timestamp"))).first()[0]

print(f"The date of the latest order is: {latest_order_date}")
```

Aggregating and feature engineering to ceate a suiatble dataframe for churn prediction

```python
# Group by customer ID and aggregate features
churn_raw_df = customer_churn_df.groupBy("cust_id", "state", "city").agg(
    F.avg("cust_value").alias("cust_value"),
    F.sum("order_value").alias("total_order_value"),
    F.avg("order_value").alias("avg_order_value"),
    F.sum("item_value").alias("total_item_value"),
    F.avg("item_value").alias("avg_item_value"),
    F.sum("quantity").alias("total_quantity"),
    F.avg("quantity").alias("avg_quantity"),
    F.avg("unit_price").alias("avg_unit_price"),
    F.count("order_id").alias("total_orders"),
    F.countDistinct("p_id").alias("unique_products"),
    F.count(F.when(customer_churn_df.status == "COMPLETE", customer_churn_df.order_id)).alias("complete"),
    F.count(F.when(customer_churn_df.status == "PENDING", customer_churn_df.order_id)).alias("pending"),
    F.count(F.when(customer_churn_df.status == "PROCESSING", customer_churn_df.order_id)).alias("processing"),
    F.count(F.when(customer_churn_df.status == "CLOSED", customer_churn_df.order_id)).alias("closed"),
    F.count(F.when(customer_churn_df.status == "ON_HOLD", customer_churn_df.order_id)).alias("on_hold"),
    F.count(F.when(customer_churn_df.status == "PAYMENT_REVIEW", customer_churn_df.order_id)).alias("pay_review"),
    F.count(F.when(customer_churn_df.status == "SUSPECTED_FRAUD", customer_churn_df.order_id)).alias("fraud"),
    F.count(F.when(customer_churn_df.status == "CANCELED", customer_churn_df.order_id)).alias("canceled"),
    F.count(F.when(customer_churn_df.status == "PENDING_PAYMENT", customer_churn_df.order_id)).alias("pay_pending"),
    F.last("year").alias("last_order_year"),
    F.last("month").alias("last_order_month"),
)

churn_raw_df.show()
```

```python
churn_raw_df = churn_raw_df.withColumn("cust_value", F.round("cust_value", 2)) \
                         .withColumn("total_order_value", F.round("total_order_value", 2)) \
                         .withColumn("avg_order_value", F.round("avg_order_value", 2)) \
                         .withColumn("total_item_value", F.round("total_item_value", 2)) \
                         .withColumn("avg_item_value", F.round("avg_item_value", 2)) \
                         .withColumn("avg_quantity", F.round("avg_quantity", 2)) \
                         .withColumn("avg_unit_price", F.round("avg_unit_price", 2))
churn_raw_df.show()
```

```python
latest_order_date
```

Calculating days since last order and basing the newly introduced churn feature on it

```python
from pyspark.sql.functions import datediff, lit

# Calculate the difference in days between the latest order date and the last order date for each customer
churn_df = churn_raw_df.withColumn("days_since_last_order", datediff(lit(latest_order_date),
                                                          F.to_date(F.concat(F.col("last_order_year").cast("string"),
                                                          F.lit("-"), F.col("last_order_month").cast("string"), F.lit("-01")))))

# Define a churn threshold (90 days)
churn_threshold = 90

# Create a churn feature
churn_df = churn_df.withColumn("churn", F.when(F.col("days_since_last_order") >= churn_threshold, 1).otherwise(0))

churn_df.show()
```

```python
# Split the data into training and testing sets
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator


(training_data, test_data) = churn_df.randomSplit([0.8, 0.2], seed=1234)
```

Since the churn column is one that has been introduced in the data by us during our analysis - based on the customer's specific order dates - if we use the columns that the churn values are based on, in our analysis, the result will always be 100% accuracy, which is due to the date fetaures being present in the feature_columns (as correlation between churn and date would be 1.0). Unlike in real life situations, where churn is a mix of date as well as other features - such as product quality, service experience etc - in this dataset, the churn is purely based on days, so including the dates in our feature columns would not give us an idea of how well the rest of our features are able to generalise, hence latest_order_month, year and days_since_last_order are left out in this analysis.

```python
# Select features for the model
feature_cols = [
    "cust_value", "total_order_value", "avg_order_value", "total_item_value",
    "avg_item_value", "total_quantity", "avg_quantity", "avg_unit_price",
    "total_orders", "unique_products"
]
```

```python
# Create a vector assembler to combine features into a single vector
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
training_data = assembler.transform(training_data)
test_data = assembler.transform(test_data)

# Create a Logistic Regression model
lr = LogisticRegression(featuresCol="features", labelCol="churn")

# Train the model on the training data
lr_model = lr.fit(training_data)

# Make predictions on the test data
predictions = lr_model.transform(test_data)
```

```python
# Evaluate the model
evaluator = BinaryClassificationEvaluator(labelCol="churn")
accuracy = evaluator.evaluate(predictions)
print(f"Accuracy: {accuracy}")
```

On using all features, we get an accuracy of **52.08%**, we must perform feature
selection to obtain better results.
On reducing few status features with closed and not necessary to monitor orders, we see a slight imporvement in accuracy to **54.16%**. On omitting the status variables entirely, our accuracy rises upto **64.58%**


Let's explore how a Decision Tree model fares in comparison

```python
from pyspark.ml.classification import DecisionTreeClassifier
# Create a Decision Tree Classifier model
dt = DecisionTreeClassifier(featuresCol="features", labelCol="churn")

# Train the model on the training data
dt_model = dt.fit(training_data)

# Make predictions on the test data
predictions_dt = dt_model.transform(test_data)
```

```python
# Evaluate the model
evaluator = BinaryClassificationEvaluator(labelCol="churn")
accuracy = evaluator.evaluate(predictions_dt)
print(f"Accuracy: {accuracy}")
```

For the same feature set, our decision Tree Classifier gives us **56.59%** accuracy whereas the Logistic Regression method is significantly higher at **64.58%**

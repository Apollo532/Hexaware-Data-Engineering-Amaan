Mapping RDDs by columns

```python
categories_rdd_mapped = categories_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))  # Mapping by category_id
customers_rdd_mapped = customers_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by customer_id
departments_rdd_mapped = departments_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))  # Mapping by department_id
orders_rdd_mapped = orders_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by order_id
order_items_rdd_mapped = order_items_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[1]), x))  # Mapping by order_item_id (use x[1] for order_id)
products_rdd_mapped = products_rdd.map(lambda line: line.split(",")).map(lambda x: (int(x[0]), x))   # Mapping by product_id
```

Converting RDDs into Dataframes by splitting rows and assigning column names.

```python
from pyspark.sql import Row

# Split RDD and map it to Row objects
customers_rdd_mapped = customers_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    cust_id=int(x[0]),
    fname=x[1],
    lname=x[2],
    unknown_1=x[3],
    unknown_2=x[4],
    address=x[5],
    city=x[6],
    state=x[7],
    value=float(x[8])
))

# Convert to DataFrame
customers_df = customers_rdd_mapped.toDF()
customers_df.show()
```

```python
orders_rdd_mapped = orders_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    order_id=int(x[0]),
    timestamp=x[1],
    value=float(x[2]),
    status=x[3]
))

orders_df = orders_rdd_mapped.toDF()
orders_df.show()
```

```python
order_items_rdd_mapped = order_items_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    id=int(x[0]),
    order_id=int(x[1]),
    cust_id=int(x[2]),
    quantity=int(x[3]),
    value=float(x[4]),
    unit_price=float(x[5])
))

order_items_df = order_items_rdd_mapped.toDF()
order_items_df.show()
```

```python
departments_rdd_mapped = departments_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    dept_id=int(x[0]),
    name=x[1]
))

departments_df = departments_rdd_mapped.toDF()
departments_df.show()
```

```python
categories_rdd_mapped = categories_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    category_id=int(x[0]),
    dept_id=int(x[1]),
    category_name=x[2]
))

categories_df = categories_rdd_mapped.toDF()
categories_df.show()
```

```python
products_rdd_mapped = products_rdd.map(lambda line: line.split(",")).map(lambda x: Row(
    p_id=int(x[0]),
    category_id=int(x[1]),
    product_name=x[2],
    link=x[3]
))

products_df = products_rdd_mapped.toDF()
products_df.show()
```

## **Performing Customer Segmentation using KMeans Clustering**

Cleaning Customers_df

```python
departments_df_new = departments_df.withColumnRenamed("name", "dept_name")
departments_df_new.show()
```

Our dataframes have now been standardized in terms of column requirement and names

```python
# Join all dataframes using inner joins
sales_raw_df = customers_df_new.join(order_items_df_new, customers_df_new.cust_id == order_items_df_new.c_id, "inner") \
    .join(orders_df_new, order_items_df_new.o_id == orders_df_new.order_id, "inner") \
    .join(products_df_new, order_items_df_new.p_id == products_df_new.product_id, "inner") \
    .join(categories_df_new, products_df_new.cat_id == categories_df_new.category_id, "inner") \
    .join(departments_df_new, categories_df_new.d_id == departments_df_new.dept_id, "inner")

sales_raw_df.show()
```

```python
sales_raw_df = sales_raw_df.drop("c_id", "o_id", "p_id", "cat_id", "d_id")
sales_raw_df.show()
```

Preparing aggregated dataframe at granularity set for cust_id since our
 subject of segmentation is customers. A highly granular dataset would give in-accurate clusters, having multiple entries for one single customer

```python
# Preparing aggregated dataframe
from pyspark.sql.functions import count, sum, avg, countDistinct

customer_sales_df = sales_raw_df.groupBy("cust_id", "full_name") \
    .agg(
        count("order_id").alias("total_orders"),
        sum("order_value").alias("total_order_value"),
        avg("order_value").alias("average_order_value"),
        avg("item_value").alias("average_product_value"),
        avg("unit_price").alias("average_unit_price"),
        sum("quantity").alias("total_quantity"),
        avg("quantity").alias("avg_quantity_per_order"),
        countDistinct("category_id").alias("total_unique_categories"),
        countDistinct("product_id").alias("total_unique_products"),
        count(F.when(sales_raw_df.status == "COMPLETE", sales_raw_df.order_id)).alias("complete"),
        count(F.when(sales_raw_df.status == "PENDING", sales_raw_df.order_id)).alias("pending"),
        count(F.when(sales_raw_df.status == "PROCESSING", sales_raw_df.order_id)).alias("processing"),
        count(F.when(sales_raw_df.status == "CLOSED", sales_raw_df.order_id)).alias("closed"),
        count(F.when(sales_raw_df.status == "ON_HOLD", sales_raw_df.order_id)).alias("on_hold"),
        count(F.when(sales_raw_df.status == "PAYMENT_REVIEW", sales_raw_df.order_id)).alias("pay_review"),
        count(F.when(sales_raw_df.status == "SUSPECTED_FRAUD", sales_raw_df.order_id)).alias("fraud"),
        count(F.when(sales_raw_df.status == "CANCELED", sales_raw_df.order_id)).alias("canceled"),
        count(F.when(sales_raw_df.status == "PENDING_PAYMENT", sales_raw_df.order_id)).alias("pay_pending")

    )

customer_sales_df.show()
```

```python
# Rounding off decimal Values

from pyspark.sql.functions import round

customer_sales_df = customer_sales_df.withColumn("total_order_value", round("total_order_value", 2)) \
    .withColumn("average_order_value", round("average_order_value", 2)) \
    .withColumn("average_product_value", round("average_product_value", 2)) \
    .withColumn("average_unit_price", round("average_unit_price", 2)) \
    .withColumn("avg_quantity_per_order", round("avg_quantity_per_order", 2))

customer_sales_df.show()
```

Kmeans Clustering

Vector Assembler combines multiple columns into a single feature vector. It is part of the MLlib library and is used when we need to pass multiple input columns into a machine learning algorithm, which expects a single vector column as input.

Many machine learning algorithms (such as those for classification, regression, and clustering) expect a feature vector as input, rather than separate individual columns. VectorAssembler helps us create that feature vector from a set of individual features (columns) by combining them into a single column of vectors.

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.clustering import KMeans
from pyspark.ml.evaluation import ClusteringEvaluator

# Select features for clustering
feature_cols = [
    "total_orders", "total_order_value", "average_order_value",
    "average_product_value", "average_unit_price", "total_quantity",
    "avg_quantity_per_order", "total_unique_categories",
    "total_unique_products", "complete", "pending", "processing", "closed",
    "on_hold", "pay_review", "fraud", "canceled", "pay_pending"
]

# Create a vector assembler to combine features into a single vector column
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
customer_sales_assembled = assembler.transform(customer_sales_df)

# Train the KMeans model
kmeans = KMeans(featuresCol="features", k=3)
model = kmeans.fit(customer_sales_assembled)

# Make predictions (assign cluster IDs to each customer)
predictions = model.transform(customer_sales_assembled)

# Evaluate the clustering
evaluator = ClusteringEvaluator()
silhouette = evaluator.evaluate(predictions)
print("Silhouette with squared euclidean distance = " + str(silhouette))

# Show the clustered data
predictions.select("cust_id", "full_name", "prediction").show()
```

A silhouette score of 0.9725 indicates that the clusters formed by the KMeans algorithm
are very well separated and dense. It means that the data points within a cluster
are very similar to each other, and the data points in different clusters are
quite distinct. A higher silhouette score indicates better clustering quality,
with a score closer to 1 being ideal.
